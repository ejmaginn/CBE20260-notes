# Saturated R-32 Properties

Liquid and vapor density, enthalpy, and entropy of saturated R-32 (difluoromethane, CH₂F₂) at selected temperatures, computed using CoolProp.

```python
# Install CoolProp if it is missing (e.g., on Google Colab)
try:
    import CoolProp
except ImportError:
    %pip install -q CoolProp

import numpy as np
import pandas as pd
from CoolProp.CoolProp import PropsSI

pd.set_option('display.float_format', '{:.4f}'.format)
pd.set_option('display.max_columns', 20)
```

```python
fluid = 'R32'
temperatures = [253, 273, 283, 293, 308, 318, 328]  # K

rows = []
for T in temperatures:
    P_sat = PropsSI('P', 'T', T, 'Q', 0, fluid) / 1e5  # bar

    rho_L = PropsSI('D', 'T', T, 'Q', 0, fluid)        # kg/m³
    h_L   = PropsSI('H', 'T', T, 'Q', 0, fluid) / 1e3  # kJ/kg
    s_L   = PropsSI('S', 'T', T, 'Q', 0, fluid) / 1e3  # kJ/(kg·K)

    rho_V = PropsSI('D', 'T', T, 'Q', 1, fluid)        # kg/m³
    h_V   = PropsSI('H', 'T', T, 'Q', 1, fluid) / 1e3  # kJ/kg
    s_V   = PropsSI('S', 'T', T, 'Q', 1, fluid) / 1e3  # kJ/(kg·K)

    rows.append({
        'T (K)':          T,
        'P_sat (bar)':    P_sat,
        'ρ_L (kg/m³)':   rho_L,
        'h_L (kJ/kg)':   h_L,
        's_L (kJ/kg·K)': s_L,
        'ρ_V (kg/m³)':   rho_V,
        'h_V (kJ/kg)':   h_V,
        's_V (kJ/kg·K)': s_V,
    })

df = pd.DataFrame(rows).set_index('T (K)')
```

```python
print('Saturated R-32 Properties (CoolProp)')
print('=' * 80)
print(df.to_string())
```

```python
# Styled display for notebooks
df.style \
    .set_caption('Saturated R-32 Properties') \
    .format({
        'P_sat (bar)':    '{:.4f}',
        'ρ_L (kg/m³)':   '{:.2f}',
        'h_L (kJ/kg)':   '{:.4f}',
        's_L (kJ/kg·K)': '{:.6f}',
        'ρ_V (kg/m³)':   '{:.4f}',
        'h_V (kJ/kg)':   '{:.4f}',
        's_V (kJ/kg·K)': '{:.6f}',
    })
```

```python
# Export to CSV (opens directly in Excel)
# utf-8-sig adds a BOM so Excel recognises the encoding on both Mac and Windows
csv_path = 'R32_saturation.csv'
df.to_csv(csv_path, encoding='utf-8-sig')
print(f'Saved: {csv_path}')
```
